```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Bob'],
    'Age': [20, 21, np.nan, 22, 21],
    'Score': [85, 90, 78, np.nan, 90]
}

df = pd.DataFrame(data)

df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())

df['Score'] = df['Score'].fillna(df['Score'].mean())

df
```
